Right now columns like genres, keywords, cast, and crew contain JSON-like strings, for example:

[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}]

We need to convert them into simple words:

Action Adventure

```python
import ast

# Create a helper function
def convert(obj):
    L = []
    
    for i in ast.literal_eval(obj):
        L.append(i['name'])
        
    return L

#This function converts: [{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}]

# into: this ["Action", "Adventure"]
```

```python
movies["genres"] = movies["genres"].apply(convert)
```

```python
movies["genres"].head()
```

```python
# same for Keywords
movies["keywords"] = movies["keywords"].apply(convert)
movies["keywords"].head()
```

```python
# Extract the top 3 actors
def convert3(obj):
    if isinstance(obj, list):
        return obj[:3]
    
    return [item["name"] for item in ast.literal_eval(obj)[:3]]


movies["cast"] = movies["cast"].apply(convert3)
```

```python
def get_director(value):
    data = ast.literal_eval(value)

    for item in data:
        if item["job"] == "Director":
            return [item["name"]]

    return []


movies["crew"] = movies["crew"].apply(get_director)

movies.rename(
    columns={"crew": "director"},
    inplace=True
)
```

```python
movies.head()
```

```python
# Convert overview into a list as we want individual words
movies["overview"] = movies["overview"].apply(lambda x: x.split())
```

Remove spaces inside names

This is an important recommendation-system trick.

If we leave it like this, NLP treats:

Sam
Worthington

as separate words.

Instead we'll convert it to:

SamWorthington

```python
movies["genres"] = movies["genres"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["keywords"] = movies["keywords"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["cast"] = movies["cast"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies["director"] = movies["director"].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
movies.head()
```

```python
# Create the tags column
movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["director"]
)
movies[["title_x", "tags"]].head()
```

```python
# Convert the list into text
movies["tags"] = movies["tags"].apply(lambda x: " ".join(x))
movies[["title_x", "tags"]].head()
```

```python
# Keep only the required columns
movies = movies[[ "title_x", "tags"]]
movies.head()
```

Convert text to lowercase

This is important for consistent NLP processing.

```python
movies["tags"] = movies["tags"].apply(lambda x: x.lower())
```

```python
movies["tags"].head()
```

Install NLTK data

We are going to use stemming to make related words more similar.

```python
!py -m pip install nltk
```

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    
    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)
```

```python
movies["tags"] = movies["tags"].apply(stem)
```

```python
movies["tags"].head()
```

```python
movies.head()
```

Now we move to the actual machine-learning representation.
TF-IDF Vectorization: We need to convert the tags text into numbers.

```python
!py -m pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# We'll use 5,000 features:
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
# Transform the tags
vectors = tfidf.fit_transform(movies["tags"])
```

```python
vectors.shape
# That means: 4803 movies  × 5000 NLP features
```

Calculate Cosine Similarity

Now comes the main recommendation algorithm.

We want to answer: "How similar is Movie A to every other movie?"

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

Test the similarity

```python
# first its index
movies[movies["title_x"] == "Avatar"].index
```

```python
similarity[0]
```

Build the recommendation function

```python
def recommend(movie):
    movie_index = movies[movies["title_x"] == movie].index[0]
    
    distances = similarity[movie_index]
    
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    for i in movie_list:
        print(movies.iloc[i[0]]["title_x"])
```

```python
recommend("Avatar")
```
